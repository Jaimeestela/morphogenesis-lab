# Linear stability analysis
Computation of the dominant mode k (max Re(λ) of J − k² D).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path().resolve().parents[1] / "src"))

from morphogenesis.stability import dominant_mode
from morphogenesis.config import get_default_schnakenberg
import matplotlib.pyplot as plt

params = get_default_schnakenberg()
k_dom, lam, k_arr, re_lam = dominant_mode("schnakenberg", params, k_min=0, k_max=15, n_k=150)
plt.plot(k_arr, re_lam)
plt.axvline(k_dom, color="r", linestyle="--", label=f"k_dom={k_dom:.3f}")
plt.xlabel("k"); plt.ylabel("Re(λ)"); plt.legend(); plt.title("Schnakenberg stability")
plt.show()
```
